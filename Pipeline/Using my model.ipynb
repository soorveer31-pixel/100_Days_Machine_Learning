```python
import pickle
import numpy as np 
```

```python
ohe_sex = pickle.load(open('models/ohe_sex.pkl','rb'))
ohe_embarked = pickle.load(open('models/ohe_embarked.pkl','rb'))
clf = pickle.load(open('models/clf.pkl','rb'))
ohe_alone = pickle.load(open('models/ohe_alone.pkl','rb'))
```

```python
# Assume input
# pclass/sex/age/sibsp/parch/fare/embarked/alone 
test_input = np.array([3,'male',40,1,0,7.9,'S',True],dtype = object).reshape(1,8)
```

```python
test_input
```

```python
test_input_sex = ohe_sex.transform(test_input[:,1].reshape(1,1))
```

```python
test_input_sex
```

```python
test_input_embarked = ohe_embarked.transform(test_input[:,-2].reshape(1,1))
```

```python
test_input_age = test_input[:,2].reshape(1,1)
```

```python
test_input_alone = ohe_alone.transform(test_input[:,-1].reshape(1,1))
```

```python
test_input_transformed = np.concatenate((test_input[:,[0,3,4,5]],test_input_age,test_input_sex,test_input_embarked,test_input_alone),axis = 1)
```
