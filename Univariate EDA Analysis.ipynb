```python
import pandas as pd 
import numpy as np 
import seaborn as sns
```

```python
df = sns.load_dataset('titanic')
```

```python
df.head()
```

# 1. Categorical Data

## a. countplot

```python
sns.countplot(x = df['survived'])
```

```python
sns.countplot(x = df['pclass'])
```

```python
df['pclass'].value_counts()
```

```python
sns.countplot(x = df['sex'])
```

## b. pie chart

```python
df['sex'].value_counts().plot(kind='pie',autopct = '%.2f')
```
